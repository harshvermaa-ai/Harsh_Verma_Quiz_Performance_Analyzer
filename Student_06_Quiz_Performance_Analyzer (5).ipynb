## 4. NumPy practice
Convert one or more numeric columns to a NumPy array. Try `np.mean()`, `np.max()` and `np.min()` where appropriate.


```python
# TODO: Select numeric columns relevant to your project
# TODO: Convert them to a NumPy array using .to_numpy()
# TODO: Calculate average, maximum and minimum

# Selecting only quiz marks

quiz_data = df[['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']]
quiz_array = quiz_data.to_numpy()
print("Average Quiz Score:", np.mean(quiz_array))
print("Maximum Quiz Score:", np.max(quiz_array))
print("Minimum Quiz Score:", np.min(quiz_array))
```

## 5. Pandas analysis
Create at least two calculated columns or useful summaries. Filter and sort rows to answer questions about your dataset.


```python
# TODO: Create calculated column(s)
# TODO: Filter records using a condition
# TODO: Sort records and display the top/bottom results
# Calculate total and average marks

df['Total'] = df[['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']].sum(axis=1)

df['Average'] = df[['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']].mean(axis=1)

high_scorers = df[df['Average'] > 70]

print("Students scoring above 70:")
display(high_scorers)

sorted_students = df.sort_values(by='Average', ascending=False)

print("Students ranked by average:")
display(sorted_students)
```

## 6. Your individual project challenge
**Challenge:** Calculate average quiz score; identify top scorer and most consistent student using standard deviation.

Write your solution below and explain what the result means.


```python
# TODO: Solve your assigned challenge
# Calculate overall quiz average
overall_average = df[['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']].values.mean()

print("Overall Quiz Average:", overall_average)

# Identify top scorer
top_scorer = df.loc[df['Average'].idxmax()]

print("Top Scorer:")
print("Name:", top_scorer['Name'])
print("Student ID:", top_scorer['Student_ID'])
print("Average:", top_scorer['Average'])

# Calculate standard deviation for each student
df['Standard_Deviation'] = df[['Quiz_1', 'Quiz_2', 'Quiz_3', 'Quiz_4']].std(axis=1)

# Identify most consistent student
consistent_student = df.loc[df['Standard_Deviation'].idxmin()]

print("Most Consistent Student:")
print("Name:", consistent_student['Name'])
print("Student ID:", consistent_student['Student_ID'])
print("Standard Deviation:", consistent_student['Standard_Deviation'])
```
